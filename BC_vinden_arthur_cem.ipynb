```python
Voltage=np.array([0.11423908042471162,0.11608285076740861,0.11870774732454896,0.12236689412985173,0.13242274388634664,0.15493768450544226,0.18174585872413015,0.22537449712295982,0.266334440650601,0.3031584993913515,0.33420087676441884,0.35903334234358464,0.379075719716652,0.3952081852958178,0.40839635924967754,0.41943798414500183,0.42928952083422767,0.43804030504571045,0.46803236832324396,0.5040531321429454,0.5268082757940515,0.5587398916395525])
Resistance=np.array([3805151.450173245,3866564.981928025,2965497.58978822,3056908.565704891,2646495.021117415,1935287.882736353,1210742.4452578288,608669.51350024,313102.76842884655,173105.2354867439,103391.19755901337,67437.52963726512,47948.75996000781,36364.24781203489,28881.397448254265,23814.06818390537,20234.52106550992,17578.962057027176,10439.420186845642,5789.426814082295,4033.8577599230957,2550.6003842838704])
Current=Voltage/Resistance
plt.plot(Voltage,np.log(Current),'o')

def function(x,a,b):
    return a*x+b
x_test=np.linspace(0.1,0.6,100)
popt,pcov=curve_fit(function,Voltage,np.log(Current))
plt.plot(x_test,function(x_test,*popt),'-',label='fit: a=%5.3f, b=%5.3f' % tuple(popt)) 

print(popt[0],popt[1])
q=1.602176634e-19
n=2.0
T=298.15
Boltzmann_constant_graphical=-np.log(np.e)*q/(popt[0]*n*T)
print('Boltzmann constant from graphical method is: ', Boltzmann_constant_graphical)
print('Boltzmann constant is: ', 1.380649e-23)

error_n=0.1
error_T=0.5
error_a= pcov[0][0]**0.5

# Relative uncertainty (propagation of uncertainty)
relative_error_k = np.sqrt(
    (error_a / popt[0])**2 +
    (error_n / n)**2 +
    (error_T / T)**2
)

# Absolute uncertainty
error_k_B = abs(Boltzmann_constant_graphical) * relative_error_k

print("Boltzmann constant:", Boltzmann_constant_graphical)
print("Absolute uncertainty:", error_k_B)
print("Relative uncertainty:", relative_error_k * 100, "%")

print(f"k_B = ({Boltzmann_constant_graphical:.4e} +/- {error_k_B:.2e}) J/K")


```
